# Reproducing the results

You can check this work at three levels, depending on how much time you have.

The quickest is to run this notebook from top to bottom. It loads the saved tables and draws every figure without refitting anything. The section "Where each result comes from" lists the script that produced each one.

If you want to confirm the inputs, the Data section links to every dataset along with its DOI.

To rerun everything from scratch, download the data (about 11 GB), set up the environment following the Run guide, and run `scripts/run_all.py`. The Run times section shows how long each step took on the original machine, so you know what to expect.

## Run guide

First, put the raw data in `data/raw` (see Data above). Then open a terminal in the repository folder and set up the environment once:

```
pip install uv
uv venv --python 3.12
uv sync --all-extras
```

To run the whole pipeline:

```
uv run python scripts/run_all.py                  # steps 1 to 9
uv run python scripts/run_all.py --exploratory    # also step 10
uv run python scripts/run_all.py --from 7         # restart from step 7
uv run python scripts/run_all.py --list           # show the commands without running them
```

Each step's run time is saved to `results/timings.csv`.

The diagram shows the order. Steps side by side don't depend on each other, and the dotted line marks the optional step.

To run a single step yourself, use `uv run python scripts/<command>` for each command listed. CELLS means `Samsung30T LGHG2 Tesla2170 TUBerlinM50LT LGM50`.

| Step | What it does | Commands |
|---|---|---|
| 1 | Clean the raw data | `clean_dataset.py` |
| 2a | Check the cleaned data | `verify_cleaning.py` |
| 2b | Simulate the known-truth cell | `simulate_known_truth.py` |
| 3 | OCV curves and ARX baseline | `run_arx.py`, then `run_arx.py SimLGM50` |
| 4 | Equivalent-circuit models | `run_ecm.py`, then `run_ecm.py SimLGM50` |
| 5a | SINDy, full and no_bv libraries | `run_sindy.py`, then `run_sindy.py SimLGM50` |
| 5b | LSTM, five seeds (uses a GPU if available, resumable) | `run_lstm.py CELLS Panasonic18650PF` |
| 6 | SINDy ensemble (takes hours, resumable) | `run_sindy_ensemble.py CELLS`, `run_sindy_ensemble.py CELLS --aggregate`, `sindy_equations.py --variant full`, `sindy_equations.py --variant no_bv` |
| 7 | Single test run, hypotheses H1 to H4 | `run_test.py` |
| 8 | Pre-registered Panasonic test | `run_panasonic.py` |
| 9 | Figures and winter exposure | `plot_data_overview.py`, `plot_results.py`, `download_weather.py`, `winter_exposure.py` |
| 10 | Exploratory (optional) | `lstm_shap.py`, `lstm_coverage.py`, `conformal_coverage.py`, `ecm_neural.py`, `ecm_adaptation.py`, `eis_ecm.py` |

If a step stops partway, run it again. The LSTM and the ensemble pick up where they left off. `run_test.py` and `run_panasonic.py` only run once by design, so add `--again --force` if you need to repeat them.

## Where each result comes from

All saved results are in `results/`, and all figures are in `results/figures/`. The tables below show which script produces each output and where it's saved. Run times for every step and cell are in `results/timings.csv`, written by `scripts/run_all.py`.

### Data

| Output | Script | Files |
|---|---|---|
| Cleaned 1 s data and manifest | `clean_dataset.py` | `data/processed/<cell>/<T>/*.parquet`, `data/processed/manifest.csv`, `data/processed/reference_capacity.json` |
| Hourly winter weather, three stations | `download_weather.py` | `data/raw/weather/<station>/<station>_<yyyy-mm>.csv`, `data/raw/weather/weather_winters_2015_2025.csv` |
| Cleaning checks (15 PASS/FAIL lines) | `verify_cleaning.py` | printed only |
| Data coverage figure | `plot_data_overview.py` | `results/figures/data_coverage_by_cell.png` |
| Usable hours by temperature figure | `plot_data_overview.py` | `results/figures/data_hours_by_temperature.png` |

### Models

These are fitted and tuned on the training temperatures.

| Output | Script | Files |
|---|---|---|
| OCV curve per cell | `run_arx.py` | `results/arx/<cell>/ocv.json` |

| LSTM weights and scores | `run_lstm.py` | `results/lstm/<cell>/h<H>_l<L>_s<seed>.pt` and `.json`, `grid.csv`, `results/lstm/summary.csv` |

## Setup

````python
from pathlib import Path

import pandas as pd
from IPython.display import Image, Markdown, display

R = Path("results")
F = R / "figures"


def show(name, width=700):
    path = F / name
    if not path.exists():
        display(Markdown(f"missing: {path}"))
        return
    display(Markdown(f"**{name}**"))
    display(Image(str(path), width=width))
````

## Run times

````python
t = pd.read_csv(R / "timings.csv")
t = t[t.ok].drop_duplicates(["step", "script", "cell", "args"], keep="last")  # latest successful run of each job

minutes = t.groupby(["step", "script"], sort=False).seconds.sum().div(60).round(1)
display(minutes.rename("minutes").to_frame())
print(f"total: {t.seconds.sum() / 3600:.1f} h")
````

How long each model took to fit and score on each cell. This is the training cost used in the model comparison.

````python
MODELS = {
    "run_arx.py": "ARX",
    "run_ecm.py": "ECM",
    "run_sindy.py": "SINDy",
    "run_lstm.py": "LSTM (5 seeds)",
    "run_sindy_ensemble.py": "SINDy ensemble",
}

per_cell = t[t.script.isin(MODELS) & t.cell.notna()]  # skip whole-run jobs like --aggregate
per_cell = per_cell.assign(model=per_cell.script.map(MODELS), minutes=per_cell.seconds / 60)
display(per_cell.pivot_table(index="cell", columns="model", values="minutes").round(1))
````
